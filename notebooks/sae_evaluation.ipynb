## Down 2.0

### Reconstruction

#### Low-Resolution (512x512)

```python
down20_df = pd.read_csv(
    "../results/sae_evaluation/reconstruction/assessment_20250822_204604/unet_down_blocks_2_attentions_0/step_488282/results.csv"  # noqa:E501
)
down20_df
```

```python
down20_df.drop(
    columns=["sae_name", "sae_path", "target_module", "avg_timestep_r2"],
    inplace=True,
)
```

```python
# Compute mean and std for every metric

mean_values = down20_df.mean()
std_values = down20_df.std()

formatted_results_down20 = {}
for col in down20_df.columns:
    if "r2" in col:
        # Format as percentage with 2 decimal places
        formatted_results_down20[col] = (
            f"{mean_values[col] *100:.2f} ± {std_values[col] * 100:.2f}%"
        )
    else:
        formatted_results_down20[col] = (
            f"{mean_values[col]:.3f} ± {std_values[col]:.3f}"
        )

# Create DataFrame for display
down20_res = pd.DataFrame([formatted_results_down20])
down20_res
```

#### High-Resolution (768x768)

```python
down20_df_high = pd.read_csv(
    "../results/sae_evaluation/reconstruction/assessment_20250825_162118/unet_down_blocks_2_attentions_0/step_488282/results.csv"  # noqa:E501
)
```

```python
down20_df_high.drop(
    columns=["sae_name", "sae_path", "target_module", "avg_timestep_r2"],
    inplace=True,
)
```

```python
mean_values = down20_df_high.mean()
std_values = down20_df_high.std()

formatted_results_down20_high = {}
for col in down20_df.columns:
    if "r2" in col:
        # Format as percentage with 2 decimal places
        formatted_results_down20_high[col] = (
            f"{mean_values[col] *100:.3f} ± {std_values[col] * 100:.3f}%"
        )
    else:
        # Format with 6 decimal places for other metrics
        formatted_results_down20_high[col] = (
            f"{mean_values[col]:.3f} ± {std_values[col]:.3f}"
        )

down_20_res_high = pd.DataFrame([formatted_results_down20_high])
down_20_res_high
```

### Knockout

```python
down20_knockout_df = pd.read_csv(
    "../results/sae_evaluation/feature_removal/assessment_20250823_162012/unet_down_blocks_2_attentions_0/step_488282/results.csv"
)  # noqa: E501
metrics = ["lpips", "mean_manhattan", "median_manhattan", "fid"]
intervals = ["0-9", "10-19", "20-24"]

summary = []
for interval in intervals:
    for metric in metrics:
        col = f"removal_{interval}_{metric}"
        summary.append(
            {
                "interval": interval,
                "metric": metric,
                "mean": down20_knockout_df[col].mean(),
                "std": down20_knockout_df[col].std(),
                "mean±std": f"{down20_knockout_df[col].mean():.3f} ± {down20_knockout_df[col].std():.3f}",  # noqa: E501
            }
        )

down20_knockout_results_df = pd.DataFrame(summary)
print(
    down20_knockout_results_df.pivot(
        index="metric", columns="interval", values="mean±std"
    )
)
```
